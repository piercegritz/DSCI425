## Problem 1: 
1.	Load the data and use pair plots to visualize the relationships.  Identify a few features that look like promising discriminators. (breast data)
2.	Split the data into training and test sets.
3.	Use the training data to find the best classifier and report on the winning model.
4.	Evaluate the performance of the winning model using the test set.


```python
(breast := pl.read_csv("data/BreastDiag.csv"))
```

# Pairplot

```python
(X_breast :=
 breast
 .drop('Diagnosis')
 .to_pandas()
)
```

```python
(y_breast :=
 breast
 .select('Diagnosis')
 .to_numpy()
 .ravel()
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_breast, y_breast, test_size=0.2, random_state=1023, stratify=y_breast)
```

```python



(generic_cls1 := Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("classifier", None)
]))
```

```python
(log_reg_grid :=
 {'classifier': [LogisticRegression(max_iter=10000)
               ]
 }
)
```

```python
(bayes_grid := 
    {'classifier': [GaussianNB()
               ]
 })
```

```python
(QDA_grid := 
    {'classifier': [QuadraticDiscriminantAnalysis()
               ]
 }
)
```

```python
(LDA_grid := 
    {'classifier': [LinearDiscriminantAnalysis()
               ]
 }
)
```

```python
std_weights = ['uniform', 'distance']
std_metrics = ['minkowski', 'euclidean', 'manhattan']
```

```python
(knn := {
    'classifier': [KNeighborsClassifier()],
    'classifier__n_neighbors': list(range(3, 12, 2)),
    'classifier__weights': std_weights,
    'classifier__metric': std_metrics,
    'classifier__p': [1, 2],  
})
```

```python
(combined_grids1 := [log_reg_grid, bayes_grid, QDA_grid, LDA_grid, knn])
```

```python
(folds := StratifiedKFold(n_splits=5, shuffle=True, random_state=1204))
```

```python
(combined_grid_search1 := GridSearchCV(
    generic_cls1,
    combined_grids1,
    cv=folds,
    scoring='balanced_accuracy',
    n_jobs=-1,
    verbose=1
))
```

```python
combined_grid_search1.fit(X_train, y_train)
```

```python
metrics = ['accuracy',
           'balanced_accuracy',
           'f1_micro',
           ]
```

```python


(cv_test_scores :=
    cross_validate(combined_grid_search1, X_train, y_train,
               cv=folds,
               scoring=metrics,
               verbose=1,
               n_jobs=-1,
               )
    >> rec.subset([f'test_{m}' for m in metrics])
    >> rec.map(np.mean)
)
```

```python
combined_grid_search1.best_params_
```

```python
(y_pred := combined_grid_search1.predict(X_test))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay


ConfusionMatrixDisplay.from_predictions(y_test,
                                        y_pred,
                                        display_labels=["B", "M"]
                                       )
```

```python
from sklearn.metrics import RocCurveDisplay

y_pred_proba = combined_grid_search1.predict_proba(X_test)[:, 1]
RocCurveDisplay.from_predictions(y_test, y_pred_proba, pos_label='M')
```

<font color = "orange">
The winning model for the breast data was the logistic regression model with data standardization to predict breast cancer diagnosis (benign vs. malignant). Using 5-fold cross-validation with balanced accuracy as the evaluation metric, the model achieved perfect classification on the test set with an AUC of 1.00. The confusion matrix and ROC curve demonstrate excellent performance with hardly any misclassifications on the held-out test data.
</font>

```python
(train2 := pl.read_csv("data/genre_train.csv"))
```

```python
(X_music_train :=
 train2
 .drop('GENRE')
 .to_pandas()
)
```

```python
(y_music_train := 
 train2
 .select('GENRE')
 .to_numpy()
 .ravel())
```

```python
(test2 := pl.read_csv("data/genre_test.csv"))
```

```python
(generic_cls2 :=
 Pipeline(steps = [
     ('scaler', StandardScaler()),
     ('classifier', None)  
 ])

)
```

```python

(log_reg_grid :=
 {'classifier': [LogisticRegression(max_iter=10000),
                 OneVsRestClassifier(LogisticRegression(max_iter=10000)),
                 OneVsOneClassifier(LogisticRegression(max_iter=10000))
               ],
 }
)
```

```python
(bayes_grid := 
    {'classifier': [GaussianNB(),
                 OneVsRestClassifier(GaussianNB()),
                  OneVsOneClassifier(GaussianNB())
               ],
 })
```

```python
(QDA_grid := 
    {'classifier': [QuadraticDiscriminantAnalysis(),
                 OneVsRestClassifier(QuadraticDiscriminantAnalysis()),
                  OneVsOneClassifier(QuadraticDiscriminantAnalysis())
               ],
 }
)
```

```python
(LDA_grid := 
    {'classifier': [LinearDiscriminantAnalysis(),
                 OneVsRestClassifier(LinearDiscriminantAnalysis()),
                  OneVsOneClassifier(LinearDiscriminantAnalysis())
               ],
 }
)
```

```python
std_weights = ['uniform', 'distance']
std_metrics = ['minkowski', 'euclidean', 'manhattan']
```

```python
(knn_alone := {
    'classifier': [KNeighborsClassifier()],
    'classifier__n_neighbors': list(range(3, 12, 2)),
    'classifier__weights': std_weights,
    'classifier__metric': std_metrics,
    'classifier__p': [1, 2],  
})
```

```python
(wrapped_knn := {
    'classifier': [
        OneVsRestClassifier(KNeighborsClassifier()),
        OneVsOneClassifier(KNeighborsClassifier())
    ],
    'classifier__estimator__n_neighbors': list(range(3, 12, 2)),
    'classifier__estimator__weights': std_weights,
    'classifier__estimator__metric': std_metrics,
    'classifier__estimator__p': [1, 2],
})
```

```python
(knn_full_grid := [knn_alone, wrapped_knn])
```

```python
(combined_grids2 := [log_reg_grid, bayes_grid, QDA_grid, LDA_grid] + knn_full_grid)
```

```python
(combined_grid_search2 := GridSearchCV(
    generic_cls2,
    combined_grids2,
    cv=folds,
    scoring='balanced_accuracy',
    n_jobs=-1,
    verbose=1  ))
```

```python
combined_grid_search2.fit(X_music_train, y_music_train)
```

```python
(cv_test_scores :=
    cross_validate(combined_grid_search2, X_music_train, y_music_train,
               cv=folds,
               scoring=metrics,
               verbose=1,
               n_jobs=-1,
               )
    >> rec.subset([f'test_{m}' for m in metrics])
    >> rec.map(np.mean)
)
```

```python
combined_grid_search2.best_params_
```

```python
combined_grid_search2.best_estimator_
```
